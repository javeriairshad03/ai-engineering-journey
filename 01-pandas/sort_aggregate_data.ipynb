# **SORT, AGGREGATE**

#### **1. SORTING** using pandas
1. **sorting single column:**
    - .sort_values(by='ColumnName', True/False, inplace='True')

2. **sorting multiple columns:**
    - .sort_values(by=['Column1', 'Column2'], ascending=['True/False', 'True/False'], inplace=True)


```python
import pandas as pd

data = {
    "Name": ["Sara", None, "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra"],
    "Age": [18, None, 28, 40, 30, 32, 29, 25]
}

df = pd.DataFrame(data)
print("Un-sorted Data")
print(df)

df.sort_values(by='Age', ascending=False, inplace=True)
print("\nAfter Sorting by Age")
print(df)
```
