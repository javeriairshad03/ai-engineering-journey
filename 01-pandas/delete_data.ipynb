```python
import pandas as pd

data = {
    "Name": ["Sara", "Ali", "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra"],
    "Age": [18, 19, 28, 40, 30, 32, 29, 25],
    "Salary": [50000, 60000, 23000, 60000, 45000, 62000, 40000, 53000],
    "Performance_score": [80, 90, 78, 82, 88, 92, 89, 77]
}

df = pd.DataFrame(data)
print(df)
print()

# to delete the desired row
delete_last = df.drop(7)
print(delete_last)
print()

# to delete the desired column
df.drop(columns=["Performance_score"], inplace=True)
print(df)
```

```python
# delete duplicates
data = {
    "Name": ["Sara", "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra", "Sara"],
    "Age": [18, 28, 40, 30, 32, 29, 25, 18],
    "City": ["Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Salary": [50000, 23000, 60000, 45000, 62000, 40000, 53000, 50000]
}

dataframe = pd.DataFrame(data)

# .duplicated() ---> checks if there's any clone of whole row
print(f"Duplicate row(s) found: {dataframe.duplicated().sum()}")

# to drop duplicates, and just keeping the first occurence
dataframe = dataframe.drop_duplicates()

print(dataframe)
```
