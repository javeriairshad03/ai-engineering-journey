### 🐼 **Pandas for AI Engineering: Summary**

This folder contains my practice tasks/labs for mastering Data Manipulation. In AI Engineering, **80% of the work is Data Preprocessing**. If the data is "dirty," the model will fail. Below is a summary of the core concepts I have mastered.

---

#### **1. The Fundamentals (Data Exploration)**
*Before building a model, we must understand the "shape" and "health" of the data.*

* **Inspecting Data:**  `.head()` / `.tail()`: Quick look at the first/last rows.
    * `.info()`: Check for data types and non-null counts.
    * `.describe()`: Statistical summary (Mean, Std Dev, Min/Max).
* **Selection & Filtering:**  `.loc[]` and `.iloc[]`: Crucial for separating **Features ($X$)** from the **Target ($y$)**.
* **Unique Values:** * `.unique()` and `.value_counts()`: Understanding category distribution.

---

#### **2. Data Cleaning (The "Make or Break" for ML)**
*AI models are like engines; "dirty fuel" (bad data) causes them to crash.*

* **Handling Missing Values:**
    * `.isnull().sum()`: Identifying the gaps.
    * `.dropna()`: Removing incomplete records.
    * `.fillna()`: "Imputing" data (e.g., filling missing ages with the `.mean()`).
* **Removing Duplicates:**
    * `.duplicated()` & `.drop_duplicates()`: Preventing the model from "over-learning" repeating data.

---

#### **3. Intermediate (Data Transformation)**
*Preparing raw data into a format that mathematical algorithms can understand.*

* **Encoding Categorical Data:** ML models only understand numbers.
    * **Map:** `df['gender'].map({'male': 0, 'female': 1})` (Manual).
    * **One-Hot Encoding:** `pd.get_dummies(df)` (Creates binary columns for categories).
    * **Category Codes:** `df['col'].astype('category').cat.codes` (Automated).
* **Grouping & Aggregation:**
    * `.groupby()`: Analyzing behavior across segments (e.g., Avg Salary per Education Level).
* **Outlier Handling:** Using logic to filter out "noise" that might confuse the model.

---

#### **4. Advanced Manipulation (Feature Engineering)**
*Creating better features often beats using better algorithms.*

* **Creating New Features:** Deriving columns (e.g., `BirthYear` → `Age`).
* **Merging & Joining:** * `pd.merge()` & `pd.concat()`: Combining multiple data sources into one master "Training Set."
* **Apply Function:** * `.apply(lambda x: ...)`: Running custom logic across every row to transform data specifically for the model's needs.
