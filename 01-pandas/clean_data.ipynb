### **Cleaning and Handling Missing Data**

1. **NaN:** (not a number) - by default python treat as NaN
2. **None:** (for object data type)

Some common methods:
- **isnull()** ---> True - (missing) OR False - (have some value)
- **dropna()** ---> requires 2 parameters - (axis, inplace) ---- [axis=0] rows, [axis=1] columns
- **fillna()** ---> requires 2 parameters - (values, inplace)  
- **interpolate()**

```python
import pandas as pd

data = {
    "Name": ["Sara", None, "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra"],
    "Age": [18, None, 28, 40, 30, 32, 29, 25],
    "Salary": [50000, None, 23000, 60000, 45000, 62000, 40000, 53000],
    "Performance_score": [80, None, 78, 82, 88, 92, 89, 77]
}

df = pd.DataFrame(data)
print(df)
print()

# 1. detecting missing values
print(df.isnull())

print()
print("To know how many values missing?")

print(df.isnull().sum())

```

```python
# 2. after detection, now we have to handle the data ---> either remove that, or fill it with some data

# fillna() : for filling missing data

df['Age'] = df['Age'].fillna(df['Age'].mean(), inplace=True)
print(df)

```

```python
# 3. interpolate() ---> have methods = linear, polynomial, time

data = {
    "Time" : [1, 2, 3, 4, 5],
    "Value" : [10, None, 30, None, 50]
}

df = pd.DataFrame(data)
print("Before interpolation")
print(df)


df['Value'] = df["Value"].interpolate(method="linear", axis=0, inplace=True)
print("After interpolation")
print(df)
```
