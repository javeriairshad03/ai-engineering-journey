#### 🔄 **Encoding Comparison: Choosing the Right Method**

| Method | Best For... | Complexity | Tool / Library |
| :--- | :--- | :--- | :--- |
| **`.map()`** | Small, specific lists (e.g., Male/Female) | **High** (Manual) | Pandas |
| **`get_dummies`** | Categories with no "order" (Colors, Cities) | **Low** (Auto) | Pandas |
| **`.cat.codes`** | Fast, internal integer coding | **Lowest** (Auto) | Pandas |
| **`LabelEncoder`** | Real Machine Learning pipelines | **Medium** (Pro) | Scikit-Learn |


#### **Quick Notes:**
1. **Got a small list?** Use `.map()`. It’s explicit and clear.
2. **Got many categories with no rank?** Use `pd.get_dummies()`. This is **One-Hot Encoding**, which prevents the AI from thinking "City 2" is "better" than "City 1."
3. **Just want a quick number?** Use `.cat.codes`. It’s the fastest way to turn strings into numbers inside a single notebook.
4. **Building a professional AI Model?** Use `LabelEncoder`. It allows you to "save" the encoding rules to use again later on new data.

```python
# 1. manual encoding ---> .map({"data":0, "data2": 1, ...})

import pandas as pd
data = {
    "Name": ["Sara", "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra", "Sara"],
    "Age": [18, 28, 40, 30, 32, 29, 25, 18],
    "City": ["Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Salary": [50000, 23000, 60000, 45000, 62000, 40000, 53000, 50000]
}

dataframe = pd.DataFrame(data)
dataframe['City_code'] = dataframe['City'].map({'Karachi':0, 'Lahore':1, 'Islamabad':2})
print(dataframe)
```

```python
# 2. quick but dirty encoding practise ---> .get_dummies()
import pandas as pd

data = {
    'Car_Model': ['Toyota', 'Honda', 'Ford', 'Toyota', 'Honda'],
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red'],
    'Price': [25000, 22000, 28000, 24000, 21000]
}

df = pd.DataFrame(data)
print("----- Original Data -----")
print(df)

df_encoded = pd.get_dummies(df, columns=["Car_Model", "Color"], dtype=int)
print("\n----- One-Hot Encoding -----")
print(df_encoded)
```

```python
# 3. Advanced Feature: creating a 'Seniority' flag (1 if Age > 30, else 0)

import pandas as pd
data = {
    "Name": ["Sara", "Irfan", "Usman", "Umer", "Javeria", "Wajeeha", "Bushra", "Sara"],
    "Age": [18, 28, 40, 30, 32, 29, 25, 18],
    "City": ["Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Salary": [50000, 23000, 60000, 45000, 62000, 40000, 53000, 50000]
}

df = pd.DataFrame(data)
df.drop_duplicates(inplace=True)
print(df)

df['is_Senior'] = df['Age'].apply(lambda x:1 if x>30 else 0)
print(f"\n{df}")
```
