```python
# syntax to import library, and load CSV file
import pandas as pd

df = pd.read_csv("orders.csv")
print(df)
```

```python
# df.Country or df["Country"] ---> same result, but one is accessing 'with method' and second is 'with index'
df.Country

df["Country"]
```

```python
# syntax for viewing only more than one column or desired columns
df[["Country", "Product", "Price"]]
```

```python
# index location - iloc : to view a row at a specific index
df.iloc[0]

# to access specific entry of that row
df.iloc[1]["Country"]

# if we want to convert ouput to list view ---> simply wrap it in list()
```

```python
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]

# and if we add "USA", "Sweden" ---> gives eror, as python dont process the comma in here so we use .isin()
df[
    (df["Category"] == "Electronics") & 
    (df["Country"].isin(["USA", "Sweden"]))
]
```
