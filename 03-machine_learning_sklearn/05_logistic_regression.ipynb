### **Syntax for Logistic Regression**
1. **Preparation**
    - X = [[features]], y = [0, 1]  ---> Target y must be Categorical (classes).
2. **Initialize**	
    - clf = LogisticRegression()  --->  Create the Classifier (The "Decision Maker").
3. **Train**	
    - clf.fit(X, y)	  --->   Model learns the S-Curve (Sigmoid).
4. **Confidence**	
    - prob = clf.predict_proba([[X]])	--->   CRITICAL: Gets the % chance for each class.
5. **Decide**
    - ans = clf.predict([[X]])	--->  The final label (0 or 1) based on a 0.5 threshold.
