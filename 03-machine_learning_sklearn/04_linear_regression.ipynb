### **1. Smart Home Energy Predictor**
**Problem: A company wants to predict the electricity bill (PKR) based on the number of units consumed.**

- Interview Logic: If the units are 0, the bill shouldn't be 0 because there is always a "Fixed Meter Rent."

```python
from sklearn.linear_model import LinearRegression
import numpy as np 

# Data: Units consumed vs Monthly Bill (PKR)
X = [[50], [100], [200], [300], [500]] # Units
y = [1500, 2800, 5500, 8200, 13500]    # Bill

model = LinearRegression()
model.fit(X,y)

def predict_bill(units):
    if units < 0:
        print("Invalid Input: Units cannot be negative or less than Zero.")

    bill = model.predict([[units]])[0]

    # bill has 'Fixed Rent' - upperlimit can be set by transformer's capacity
    final_bill = np.clip(bill, 500, None)
    return round(final_bill, 2)

try: 
    user_input = input("Enter the number of units consumed: ")
    units_consumed = float(user_input)

    result = predict_bill(units_consumed)

    if isinstance(result, str):
        print(result)
    else:
        print(f"Based on Consumed Units: {units_consumed} units")
        print(f"Estimated Monthly Bill: {result:,} PKR")

except ValueError:
    print("Invalid Input! Please enter a numerical value (e.g. 150).")
```

### **2. E-commerce Delivery Time Predictor**
**Problem: Predict delivery time in minutes based on distance in km.**
- Interview Logic: Even for 0.1km, a rider needs time to park and deliver (a "base time"). 
- Also, delivery can't take "negative minutes."

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
import numpy as np

X_dist = np.array([[1], [2], [3], [5], [8], [10], [12], [15]])   # Distance in km
y_time = np.array([15, 20, 25, 38, 50, 55, 70, 85])              # Time in minutes

model = LinearRegression()
model.fit(X_dist, y_time)

predict_on_train = model.predict(X_dist)
error_rate = mean_absolute_error(y_time, predict_on_train)

def estimate_delivery(km):
    if km < 0:
        raise ValueError("Distance cannot be Negative.")
    
    predictions = model.predict([[km]])[0]

    base_time = np.clip(predictions, 10, 180)       #base time: minimum - 10min , maximum - 120min(2hr)
    return int(base_time)

print("------- AI FoodDelivery Estimator -------\n")
print(f"Average Error Margin: {error_rate:.1f} minutes \n")

while True:
    user_input = input("Enter the Distance in km (or type 'exit' to quit): ").strip().lower()

    if user_input == 'exit':
        print("Closing... GoodBye!")
        break

    try:
        distance = float(user_input)
        time = estimate_delivery(distance)

        print(f"Distance: {distance} km")
        print(f"Estimated Time: {time} mins")
        print(f"Note: Expect Arrival between {time-2} - {time+2} mins")

    except ValueError as e:
        msg = str(e) if 'Distance' in str(e) else "Please enter Valid Distance (in numbers)"
        print(f"Input Error: {msg}")
```
