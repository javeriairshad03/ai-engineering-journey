```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [40, 50, 65, 75, 90]

model = LinearRegression()
model.fit(X,y)

hours_studied = float(input("Enter how many hours you studied : "))

predicted_score = model.predict([[hours_studied]])

print(f"Based on your hour(s) {hours_studied}, you many score around {predicted_score}")
```
