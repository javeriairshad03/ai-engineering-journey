#### **1. MinMaxScaler:**
     > Rescaling OR Normalization 
     > Sensitive to OUTLIERS
     > Squashes all values into a fixed 0-to-1 (or -1 to 1) box 
     > Used for data with strict boundaries (Images, Neural Networks , KNNs)

#### **2. StandardScaler:**
     > Z-score OR Standardization 
     > Handles OUTLIERS 
     > Results all values into (-3 and 3) box 
     > Default and Classic choice for most models (Linear Regression, SVM) 

#### **3. RobustScaler:**
     > Handles OUTLIERS robustly and efficiently (e.g., House Prices, Salaries) 
     > Uses Median and IQR to ignore extreme values 
     > Best for data with many "weird" or extreme values, ignores outliers and scales "normal" middle part of data 

#### **4. PowerTransformer:**
     > Gaussian Mapping 
     > Transforms "tilted" data into a Normal (Bell) Curve 
     > Best Use-Case: Highly skewed data where models require normality 

#### **5. Normalizer:**
     > Unit Vector Scaling
     > Scales each row so its length (magnitude) is 1 
     > Use Case: Text Mining (NLP) or comparing document similarities 


```python
import pandas as pd
import numpy
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler

data = {
    'Name': ['Aman', 'Priya', 'Rahul', 'Anjali', 'Ravi'],
    'Age': [25, 45, 31, 52, 28],                # Normal data
    'Salary': [50000, 60000, 55000, 10000000, 52000], # Outlier (10 Million!)
    'Credit_Score': [45, 88, 56, 92, 71]        # Fixed range 0-100
}

df = pd.DataFrame(data)

print(f"--------- Original Data --------- \n{df}")

std_scaler = StandardScaler()
df["Age_Zscore"] = std_scaler.fit_transform(df[["Age"]])

minMax__scaler = MinMaxScaler()
df["Salary_Scaled"] = minMax__scaler.fit_transform(df[["Salary"]])

robst_scaler = RobustScaler()
df["Credit_Final"] = robst_scaler.fit_transform(df[["Credit_Score"]])

print("\n--------- Scaled Results ---------")
print(df[["Name", "Age_Zscore", "Salary_Scaled", "Credit_Final"]])
```
