```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

df = pd.read_csv('sample_data.csv')

df_label = df.copy()

le = LabelEncoder()
df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])
df_label['Passed_Encoded'] = le.fit_transform(df_label['Passed'])

print("Label Encoded Data")
print(df_label[['Name', 'Gender', 'Gender_Encoded','City', 'Passed', 'Passed_Encoded']])
```

```python
# to convert the (default) True/False ---> binary (0s/1s)
# df_encoded = pd.get_dummies(df, columns=['City'], dtype=int)

df_encoded = pd.get_dummies(df, columns=['City'])
print("\n One-Hot Encoded Data (City)")
print(df_encoded)
```

### **Quick fix: to convert the whole dataframe to binary 0s/1s**
    .astype() 
1. **Cleaning Numbers: When loading CSV where a column looks like numbers, but because of one empty space or a dollar sign, Pandas loads it as a String (Object). We can't do math on strings.**

    Use Case: df['Price'] = df['Price'].astype(float)

2. **To optimize memory: Pandas uses int64 (which uses a lot of RAM). If we have column like "Age" where no number is higher than 120, you don't need all that space.**

    Use Case: df['Age'] = df['Age'].astype('int8')
